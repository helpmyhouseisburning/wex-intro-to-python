## Section 1: Expressions, Variables, and Types

In Python, you can:
- **Perform calculations** (arithmetic expressions)
- **Store values** (variables)
- **Check what type of data** you have (int, str, float, bool)

```python
# Example 1: Arithmetic Expressions
print("=== Math in Python ===")
print(10 + 5)          # Addition
print(10 - 5)          # Subtraction
print(10 * 5)          # Multiplication
print(10 / 5)          # Division (always returns float)
print(10 // 3)         # Integer division (whole number only)
print(10 % 3)          # Remainder
print(2 ** 3)          # Exponent (power)
```

```python
# Example 2: Variables - storing values
print("\n=== Variables ===")
name = "Alice"
age = 25
height = 5.7
is_student = True

print(name)
print(age)
print(height)
print(is_student)

# Use variables in calculations
print(f"\nNext year, {name} will be {age + 1} years old")
```

```python
# Example 3: Data Types
print("\n=== Data Types ===")
text = "Python"
number = 42
decimal = 3.14
flag = True

print(f"Type of '{text}': {type(text)}")        # str
print(f"Type of {number}: {type(number)}")      # int
print(f"Type of {decimal}: {type(decimal)}")    # float
print(f"Type of {flag}: {type(flag)}")          # bool
```

## Section 2: Strings - Working with Text

Strings are sequences of text. You can index them, slice them, and use many built-in methods to manipulate them.

```python
# Example 1: Indexing and Slicing Strings
print("=== String Indexing & Slicing ===")
text = "PYTHON"
print(f"Full text: {text}")
print(f"First character (index 0): {text[0]}")
print(f"Last character: {text[-1]}")
print(f"Characters 0-2: {text[0:3]}")
print(f"Every other character: {text[::2]}")
print(f"Length: {len(text)}")
```

```python
# Example 2: String Methods
print("\n=== String Methods ===")
text = "hello world"
print(f"Original: {text}")
print(f"Upper case: {text.upper()}")
print(f"Capitalize: {text.capitalize()}")
print(f"Replace: {text.replace('world', 'Python')}")
print(f"Split: {text.split()}")

# Remove whitespace
text_with_spaces = "  spaces  "
print(f"\nBefore strip: '{text_with_spaces}'")
print(f"After strip: '{text_with_spaces.strip()}'")

# Check if string contains something
print(f"\nDoes 'hello world' contain 'world'? {'world' in text}")
```

```python
# Example 3: F-Strings (Modern String Formatting)
print("\n=== F-Strings ===")
name = "Alice"
age = 25
gpa = 3.85

# F-strings allow you to embed expressions inside strings
print(f"My name is {name}")
print(f"I am {age} years old")
print(f"My GPA is {gpa:.2f}")  # .2f means 2 decimal places
print(f"Next year I'll be {age + 1} years old")
print(f"Hello {name.upper()}!")  # Can use methods inside f-strings
```

## Section 3: Collections - Lists, Tuples, Dictionaries

Python has several ways to store multiple values:
- **Lists**: Ordered, changeable, allows duplicates
- **Tuples**: Ordered, unchangeable (immutable), allows duplicates
- **Dictionaries**: Key-value pairs, unordered, changeable
- **Sets**: Unordered, no duplicates

```python
# Example 1: Lists (Changeable, Ordered)
print("=== LISTS ===")
fruits = ["apple", "banana", "orange", "grape"]
print(f"List: {fruits}")
print(f"First item: {fruits[0]}")
print(f"Last item: {fruits[-1]}")
print(f"Length: {len(fruits)}")

# Modify a list
fruits[1] = "blueberry"        # Change an item
print(f"\nAfter changing: {fruits}")

fruits.append("kiwi")           # Add to end
print(f"After append: {fruits}")

fruits.insert(0, "strawberry")  # Insert at position
print(f"After insert: {fruits}")

fruits.remove("banana")         # Remove an item (only first occurrence)
print(f"After remove: {fruits}")

# Sorting
numbers = [3, 1, 4, 1, 5, 9, 2, 6]
print(f"\nOriginal: {numbers}")
print(f"Sorted: {sorted(numbers)}")
print(f"Sum: {sum(numbers)}")
```

```python
# Example 2: Tuples (Unchangeable, Ordered)
print("\n=== TUPLES ===")
colors = ("red", "green", "blue", "yellow")
print(f"Tuple: {colors}")
print(f"First: {colors[0]}")
print(f"Last: {colors[-1]}")
print(f"Length: {len(colors)}")

# Try to modify (this will error!)
# colors[0] = "pink"  # Uncomment to see the error

# But you CAN create a new tuple
colors = colors + ("pink",)
print(f"New tuple: {colors}")

# Count and find index
traffic = ("red", "yellow", "green", "red")
print(f"\nHow many 'red'? {traffic.count('red')}")
print(f"Index of first 'yellow': {traffic.index('yellow')}")
```

```python
# Example 3: Dictionaries (Key-Value Pairs)
print("\n=== DICTIONARIES ===")
student = {
    "name": "Bob",
    "age": 20,
    "major": "Computer Science",
    "gpa": 3.7
}

print(f"Dictionary: {student}")
print(f"Name: {student['name']}")
print(f"Major: {student['major']}")

# Add a new key-value pair
student["year"] = 2
print(f"\nAfter adding year: {student}")

# Modify an existing value
student["gpa"] = 3.8
print(f"After updating GPA: {student}")

# Remove a key-value pair
del student["year"]
print(f"After deleting year: {student}")

# Check if key exists
print(f"\nIs 'name' in student? {'name' in student}")
print(f"Is 'email' in student? {'email' in student}")

# Get all keys and values
print(f"Keys: {list(student.keys())}")
print(f"Values: {list(student.values())}")
```

```python
# Example 4: Sets (No Duplicates, Unordered)
print("\n=== SETS ===")
# Sets remove duplicates automatically
numbers = {1, 2, 3, 2, 1, 4, 3, 5}
print(f"Set (duplicates removed): {numbers}")

# Add and remove items
numbers.add(6)
print(f"After adding 6: {numbers}")

numbers.remove(2)
print(f"After removing 2: {numbers}")

# Set operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print(f"\nSet A: {set_a}")
print(f"Set B: {set_b}")
print(f"Union (all): {set_a | set_b}")
print(f"Intersection (common): {set_a & set_b}")
print(f"Difference (A only): {set_a - set_b}")
```

## Section 4: Conditional Logic - Making Decisions

Use `if`, `elif`, and `else` to make your program do different things based on conditions.

```python
# Example 1: Simple if/else
print("=== Simple if/else ===")
age = 18
if age >= 18:
    print("You are an adult")
else:
    print("You are not an adult")
```

```python
# Example 2: if/elif/else for grading
print("\n=== Grade Calculator ===")
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score: {score} -> Grade: {grade}")
```

```python
# Example 3: Logical operators (and, or, not)
print("\n=== Logical Operators ===")
age = 25
income = 50000
is_student = False

# AND - both conditions must be true
if age >= 18 and income >= 30000:
    print("✓ Loan approved! (Both conditions met)")
else:
    print("✗ Loan denied")

# OR - at least one condition must be true
if age < 18 or income < 20000:
    print("Not qualified for special offer")
else:
    print("✓ You qualify for special offer!")

# NOT - reverses the condition
if not is_student:
    print("You are not a student")
```

## Section 5: Loops - Repeating Code

Loops let you do the same thing many times without writing the code repeatedly.

```python
# Example 1: For loop with range()
print("=== For Loops ===")
print("Countdown from 5:")
for i in range(5, 0, -1):  # Start, Stop, Step
    print(i)
print("Blastoff! 🚀")
```

```python
# Example 2: Loop through a list
print("\n=== Loop Through Collections ===")
fruits = ["apple", "banana", "orange"]
for fruit in fruits:
    print(f"I like {fruit}")

# Loop through a dictionary
print("\nStudent grades:")
grades = {"Alice": 95, "Bob": 87, "Carol": 92}
for name, grade in grades.items():
    print(f"{name}: {grade}")
```

```python
# Example 3: While loop
print("\n=== While Loop ===")
count = 1
while count <= 5:
    print(f"Count: {count}")
    count = count + 1  # Must update the counter!
print("Done!")

# Example 4: break and continue
print("\n=== Break and Continue ===")
for i in range(10):
    if i == 3:
        continue  # Skip 3
    if i == 7:
        break     # Stop at 7
    print(i, end=" ")
print("\n(Skipped 3, stopped at 7)")
```

## Section 6: Functions - Reusable Code Blocks

Functions help you write code once and reuse it many times. Use `def` to define a function.

```python
# Example 1: Simple function
print("=== Functions ===")

def greet(name):
    """A simple greeting function"""
    print(f"Hello, {name}!")

greet("Alice")
greet("Bob")
greet("Carol")
```

```python
# Example 2: Functions with return values
print("\n=== Functions with Returns ===")

def add(a, b):
    """Add two numbers and return the result"""
    return a + b

result = add(5, 3)
print(f"5 + 3 = {result}")

# Function with multiple returns
def get_grade(score):
    """Convert score to letter grade"""
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "F"

print(f"Score 95 = Grade {get_grade(95)}")
print(f"Score 75 = Grade {get_grade(75)}")
```

```python
# Example 3: Default parameters
print("\n=== Default Parameters ===")

def introduce(name, age=18, city="Unknown"):
    """Introduce a person with optional info"""
    print(f"{name} is {age} years old and lives in {city}")

introduce("Alice")                              # Uses defaults
introduce("Bob", 25)                            # Override age
introduce("Carol", 30, "New York")              # Override both
```

## Section 7: 🎯 Mini Project - Student Grade Analyzer

Let's combine everything you learned: dictionaries, loops, conditionals, and functions!

```python
def calculate_average(scores):
    """Calculate average of scores"""
    return sum(scores) / len(scores)

def get_letter_grade(average):
    """Convert average to letter grade"""
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    else:
        return "F"

def print_report(students_data):
    """Print a grade report"""
    print("=" * 50)
    print("STUDENT GRADE REPORT")
    print("=" * 50)
    
    for name, scores in students_data.items():
        avg = calculate_average(scores)
        grade = get_letter_grade(avg)
        print(f"{name}: {avg:.2f} -> {grade}")
    
    print("=" * 50)

# Sample data
students = {
    "Alice": [95, 92, 98],
    "Bob": [87, 85, 90],
    "Carol": [78, 82, 80],
    "David": [65, 70, 68]
}

# Run the report
print_report(students)
```
